# Step 1: EEG basics

1. **EEG** is voltage from big groups of neurons, recorded as an array of `(channels, samples)`.
2. **Sampling rate** is samples per second; you can only see frequencies up to half of it (the Nyquist limit).
3. **Brain activity** is a mix of waves at different speeds, grouped into bands: delta, theta, alpha, beta, gamma.
4. **Seizures** look rhythmic, large and spreading. ML either detects them (ictal) or forecasts them (preictal vs interictal).

## Frequency bands

| Band | Range | Associated with |
| --- | --- | --- |
| Delta | 0.5–4 Hz | Deep sleep |
| Theta | 4–8 Hz | Drowsiness |
| Alpha | 8–13 Hz | Relaxed, eyes closed |
| Beta | 13–30 Hz | Alert, focused |
| Gamma | 30+ Hz | Processing |

## Seizure states

| State | Meaning |
| --- | --- |
| Preictal | Window before a seizure |
| Ictal | During the seizure |
| Postictal | Recovery afterwards |
| Interictal | Baseline between seizures |

## Check questions

**Shape of a 1-hour, 23-channel recording at 256 Hz.** The sampling rate is not an axis; it sets how many samples one second becomes.

```python
import numpy as np

FS = 256
channels = 23
seconds = 60 * 60

recording = np.zeros((channels, seconds * FS))
print("raw recording:", recording.shape)

# windowed into 1-second chunks, ready for a model
windows = recording.reshape(channels, seconds, FS).transpose(1, 0, 2)
print("1 s windows:  ", windows.shape)
```

**Why 256 Hz can't see 200 Hz.** You need at least 2 samples per cycle, so the ceiling is FS / 2 = 128 Hz (the Nyquist limit). A 200 Hz wave would alias to a fake 256 − 200 = 56 Hz.
